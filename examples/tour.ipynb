# A tour of inkorbit

Orbital diagrams in the style of 1960s black-and-white technical print. Every figure here is a `Plate`, which displays itself as the last line of a cell.

```python
import numpy as np

import inkorbit as ink
```

## Building blocks

Every function in `inkorbit.draw` is also a method on a `Plate`. Labels are placed automatically.

```python
orbit = ink.Elements(a=2.6, e=0.55, nu=np.radians(110))  # Earth radii, radians

plate = ink.Plate("Figure 1.1-1  Elliptical orbit about the Earth")
plate.body(radius=1.0, orbit=orbit)
plate.orbit(orbit)
plate.apsides(orbit, body_radius=1.0)
plate.radius_vector(orbit, body_radius=1.0)
plate.velocity_vector(orbit, length=1.0)
plate.spacecraft(orbit, label="spacecraft")
plate.true_anomaly(orbit, radius=1.5)
plate
```

## A globe

`plate.body(..., body="Mars", surface=True)` draws a planet with its surface features in stipple.

```python
plate = ink.Plate("Figure 2.1-1  Mars", page=(90, 90))
plate.body(1.0, body="Mars", surface=True, sun=None, spin=np.radians(-110), tilt=np.radians(20), label="MARS")
plate
```

## Oblique view

`view=(elevation, azimuth)` in radians shows an orbit from off its plane.

```python
molniya = ink.Elements(
    a=16_000, e=0.55, i=np.radians(63.4), raan=np.radians(40), argp=np.radians(270), nu=np.radians(150)
)
ink.diagram("orbit", molniya, view=np.radians([20, -20]), caption="Figure 3.2-2  Inclined orbit")
```

## Printed mode

`printed=True` turns the figure into a two-colour bitmap that looks scanned from an old report: ink spread, rough edges, grain and a slight skew. It is seeded, so it prints the same way every time.

```python
hohmann_plate(printed=True)
```

Pass an `ink.Printed(...)` to tune the wear.

```python
hohmann_plate(printed=ink.Printed(skew_deg=0.6, speck_density=0.03, dropout_density=0.6))
```
